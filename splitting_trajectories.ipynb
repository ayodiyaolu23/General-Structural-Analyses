```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
%matplotlib inline
from functools import reduce
import seaborn as sns
```

```python
import MDAnalysis as mda
import os
```

```python
# File paths
base_path = "/Users/ayodi/OneDrive/Documents/gaba_mutate/traj_files/wt_apo"
topology_file = os.path.join(base_path, "step5_input.psf")
trajectory_file = os.path.join(base_path, "before_gamd.dcd")
```

```python
# Load universe
u = mda.Universe(topology_file, trajectory_file)

# Total frames and chunk size
n_frames = len(u.trajectory)
chunk_size = n_frames // 4

# Output file naming
output_template = os.path.join(base_path, "before_gamd_part_{}.dcd")
```

```python
# Output file template
output_template = os.path.join(base_path, "before_gamd_part_{}.dcd")

# Split manually using a counter
for i in range(4):
    start = i * chunk_size
    end = (i + 1) * chunk_size if i < 3 else n_frames  # Ensure we cover all frames

    output_file = output_template.format(i + 1)
    with mda.Writer(output_file, n_atoms=u.atoms.n_atoms) as w:
        for ts_index, ts in enumerate(u.trajectory):
            if start <= ts_index < end:
                w.write(u.atoms)

    print(f"Saved: before_gamd_part_{i+1}.dcd (frames {start+1}–{end})")
```
